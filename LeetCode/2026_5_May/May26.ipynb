```python
def numberOfSpecialChars(word):
        lower_chars = set()
        upper_chars = set()

        for ch in word:
            if ch.islower():
                  lower_chars.add(ch)
            elif ch.isupper():
                  upper_chars.add(ch.lower())
        
        print(lower_chars & upper_chars)
        return len(lower_chars & upper_chars)

word = "AbsolutelyAmazingIDontKnowHowYouDoit"
print(numberOfSpecialChars(word))
```
